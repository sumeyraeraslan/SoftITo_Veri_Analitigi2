#### İSTANBUL VERİLERİNİ GETİRME 
df.head() verilerine baktığımızda çoğu il büyük harfle yazılmış. Araya kaynayan küçük harfli iller olabileceğinden dolayı str.upper parametresini kullandım.


```python
df_istanbul_verilerini_getirme = df[df['kurum_ili'].str.upper() == 'İSTANBUL'].copy()
```

#### NULL DEĞER KONTROLÜ

```python
kurum_ili_kontrol = df[df['kurum_ili'].isnull()]
kurum_ili_kontrol.head()
```

```python
df_il_bos = df[df['kurum_ili'].isnull() & df['kurum_ilcesi'].notnull()] #kurum ilinin boş olup kurum ilçesinin dolu olduğu senaryo mümkün mü
df_il_bos.head()
```

#### NULL DEĞER İÇERMEYEN VERİLERİMİZ NE KADAR?

```python
dolu_il_sayisi = df['kurum_ili'].notnull().sum()
print("Dolu olan il sayısı:", dolu_il_sayisi)
```

```python
dolu_il_sayisinin_kaci_istanbul = df_istanbul_verilerini_getirme['kurum_ili'].notnull().sum()
print("İstanbul dolu veri:", dolu_il_sayisinin_kaci_istanbul)
```

#### TÜRKİYE VE İSTANBUL BAZLI HASTALIK TAKİP SIKLIĞI ORTALAMASI

```python
ortalama_takip_suresi_istanbul = df_istanbul_verilerini_getirme.groupby('hastalik_adi')['takip_sikligi'].mean()
print("Hastalık bazında ortalama takip süresi:", ortalama_takip_suresi_istanbul)
```

```python
ortalama_takip_suresi_turkiye_geneli = df.groupby('hastalik_adi')['takip_sikligi'].mean()
print("Hastalık bazında ortalama takip süresi:", ortalama_takip_suresi_turkiye_geneli)
```

#### TÜRKİYE VE İSTANBUL BAZLI HASTALIKLARIN YÜZDELİK OLARAK KARŞILAŞTIRILMASI

```python
tr_takip = (df['hastalik_adi'].value_counts(normalize=True) * 100).reset_index()
tr_takip.columns = ['hastalik_adi', 'yuzde']
print(tr_takip)
```

```python
ist_takip = (df_istanbul_verilerini_getirme['hastalik_adi'].value_counts(normalize=True) * 100).reset_index()
ist_takip.columns = ['hastalik_adi', 'yuzde']
print(ist_takip)
```

Aşağıdaki çıkan sonuç verisinden elde edilebilecek sonuçlar:

- Türkiye geneline göre İstanbulda kronik böbrek, kardiyovasküler hastalıklar, astım ve obezite (izlem_kronik_bobrek, izlem_kardiyovaskuler,  izlem_astim, izlem_obezite) fazla görülmüştür.
- Bunu etkileyen sebepler başlıca büyük şehirde yaşamanın getirileri olan stres, hava kirliği, sağlıksız beslenme gibi temel sebeplerden kaynaklı olabilir.

```python
dagilim_karsilastirma = pd.merge(tr_takip, ist_takip, on='hastalik_adi', suffixes=('_turkiye', '_istanbul'))
print(dagilim_karsilastirma)
```

#### YAŞ YILLIK İSTATİSTİKLERİ 

```python
df_istanbul_verilerini_getirme['Takip_Yili'] = df_istanbul_verilerini_getirme['Takip_tarihi'].dt.year

yas_istatistikleri_yillik = df_istanbul_verilerini_getirme.groupby(['Takip_Yili', 'hastalik_adi'])['yasi'].agg(
    En_Kucuk_Yas='min', 
    En_Buyuk_Yas='max', 
    Ortalama_Yas='mean'
).reset_index()

display(yas_istatistikleri_yillik)
```

```python
pivot_ortalama_yas = yas_istatistikleri_yillik.pivot(
    index='hastalik_adi', 
    columns='Takip_Yili', 
    values='Ortalama_Yas'
)

display(pivot_ortalama_yas)
```

```python
yas_istatistikleri_yillik['Ortalama_Yas'] = yas_istatistikleri_yillik['Ortalama_Yas']
```

```python
pivot_kucuk_buyuk = yas_istatistikleri_yillik.set_index(['hastalik_adi', 'Takip_Yili']).stack().unstack('Takip_Yili')
pivot_kucuk_buyuk.index.names = ['Hastalık Adı', 'Değer']
```

Aşağıda çıkan sonuca göre ilerleyen yıllarda hastalıklarda görülen yaşın düştüğünü ve daha küçük yaşlarda görülebileceği görülmektedir. 

```python
display(pivot_kucuk_buyuk)
```

#### İSTANBULDAKİ HASTALIKLARIN HANGİ CİNSİYETTE DAHA FAZLA OLDUĞU ÇIKARIMI

İstanbul bazlı olarak bazı hastalıkların astim, diyabet, hipertansiyon,inme, kardiyovaskuler,obezite kadınlarda daha fazla görülürken koah,koronerarter, kronik bobrek erkeklerde daha fazla görülmüştür. 

```python
df_istanbul_verilerini_getirme['cBinsiyet'] = df_istanbul_verilerini_getirme['cinsiyet'].astype(str).str.strip().str.capitalize()

crosstab_df = pd.crosstab(df_istanbul_verilerini_getirme['hastalik_adi'], df_istanbul_verilerini_getirme['cinsiyet'])

display(crosstab_df)
```

#### HASTALIĞA YAKALANMADA RİSK FAKTÖRÜ VE ALIŞKANLIKLARIN ETKİSİ

```python
df_sigara = df_istanbul_verilerini_getirme.dropna(subset=['sigara_kullanimi'])
df_alkol = df_istanbul_verilerini_getirme.dropna(subset=['alkol_kullanimi'])
```

```python
sigara_oranlari = pd.crosstab(df_sigara['hastalik_adi'], df_sigara['sigara_kullanimi'], normalize='index') * 100
```

```python
alkol_oranlari = pd.crosstab(df_alkol['hastalik_adi'], df_alkol['alkol_kullanimi'], normalize='index') * 100
```

- 1.0: 'Kullanmıyor',
- 2.0: 'Kullanıyor / Düzenli Kullanıcı',
- 3.0: 'Bazen / Ara Sıra Kullanıyor',
- 4.0: 'Bırakmış / Geçmişte Kullanmış'

- Hayati risk taşıyabilecek hastalıkların sigara kullanmayı bırakması daha yüksek.
- Koah hastalarında en yüksek oranda sigara kullanımı oranı mevcutken en yüksek sigarayı bırakılması oranı da diğer hastalıklara göre daha yüksektir. Bu da sigaranın hastalığı tetikleyebilecek bir faktör olabileceğini göstermektedir.
- Koah hastalarından sonra en yüksek sigara kullanımı obezitede görülmektedir. 
- Yine aynı şekilde inme hastalarında da sigara kullanım oranın bırakılma oranından fazla oluduğu görülmektedir.Bu hastalıkta sigara kullanımının ciddi bir önem taşıdığı sonucuna varabiliriz. 

```python
display(sigara_oranlari)
```

Hastalık türüne bakmaksızın 1.0 oranı yani kullanmayan kişi sayısı daha yüksek. Bu da alkolün veri setinde hastalıkları ayrıştıran tam ve etkili bir faktör olmadığını gösteriyor.

```python
display(alkol_oranlari)
```
