## Bölüm 2

* `mount_everest_deaths` veri kümesinin kendi indeks sütunu vardır. Veriyi içe aktarırken bu mevcut indeks sütununu kullanın.  
* Hangi sütunlarda sıfır eksik değer (null) vardır?  
* Hangi sütunda en fazla eksik değer (null) var?


```python
df = pd.read_csv("mount_everest_deaths.csv", index_col=0) 
# index_col=0 parametresi, csv deki ilk sütunun indeks olarak kullanılmasını sağlar
```

```python
eksik_deger_sayilari = df.isnull().sum()
sifir_eksik_sutunlar = eksik_deger_sayilari[eksik_deger_sayilari == 0].index.tolist()
print(eksik_deger_sayilari)
print(sifir_eksik_sutunlar)
```

```python
en_fazla_eksik_sutun = eksik_deger_sayilari.idxmax()
#idxmax: en büyük değere sahip olan satırı getirir
print(en_fazla_eksik_sutun)
```
