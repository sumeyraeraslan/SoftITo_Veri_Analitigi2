## Bölüm 4

* Veri kümesinde her bir Tür'ün (Genre) kaç kez geçtiğini gösteren bir pasta (pie) grafiği oluşturun.  
* En üretken 10 yazarı bulun ve kitap sayılarını çubuk grafik (bar plot) olarak görselleştirin.  
* Kullanıcı puanlarının (rating) dağılımını gösteren bir histogram grafiği oluşturun.


```python
import matplotlib.pyplot as plt
```

```python
tur_sayilari = df['Genre'].value_counts()
```

```python
plt.figure(figsize=(8, 8))
plt.pie(tur_sayilari, labels=tur_sayilari.index, autopct='%1.1f%%', startangle=140, colors=['lightblue', 'lightgreen'])
plt.title('Kitap Türleri Dağılımı')
plt.show()
```

```python
en_uretken_10_yazar = df['Author'].value_counts().head(10)
```

```python
plt.figure(figsize=(12, 6))
en_uretken_10_yazar.plot(kind='bar', color='coral', edgecolor='black')
plt.title('En üretken 10 yazar')
plt.xlabel('Yazar')
plt.ylabel('Kitap sayısı')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 6))
plt.hist(df['User Rating'], bins=15, color='mediumpurple', edgecolor='black')
plt.title('Kullanıcı puanları (User Rating) dağılımı')
plt.xlabel('Kullanıcı puanı')
plt.ylabel('Frekans')
plt.show()
```
